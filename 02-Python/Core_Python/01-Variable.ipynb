##### Properties of Python:

1.  Python is dynamically typed.

2.  The data type of the variable in python is not defined. Its type is determined automatically once the value has been assigned to the variable.

3.  The value of the variable get changed during the runtime execution of the program and the same variable can have different types of value at different runtime execution of the program.

```python
x = 100     ####### Integer
print(x)
print(type(x))

x = 'Anish'   ##### String
print(x)
print(type(x))

x = 2+3j      ##### Complex
print(x)
print(type(x))

x = True      ##### Boolean
print(x)
print(type(x))
```

##### OBSERVATIONS:

1.  The value of the variables changes during the entire execution of the program.

2.  The datatype of the variable gets automatically defined once we assign the value to the variable.

3.  The same variable 'x' has thre different values. One with Integer, String and Complex at different instances of time.

#### Rules for defining the variable

1.  A variable must begin with an alphabet or an underscore

2.  It should not begin with a digit

3.  It must have an underscore in between if needed

4.  It should not be any keyword

5.  It should not have hyphen in between

6. It should not used any special character like ($,%,&) etc.

```python
##### Rule 1: A variable must begin with an alphabet or an underscore

name = 'Parag'
print(name)

_name = 'Anish'
print(_name)
```

```python
#### Rule 2: It should not begin with a digit

# 1name = 'Parag'(SyntaxError: invalid decimal literal)
```

```python
#### Rule 3: It must have an underscore in between if needed

first_name = 'Parag'
last_name  = 'Pujari'

print(first_name)
print(last_name)
```

```python
#### Rule 4: It should not be any keyword

#class = 'Data'(SyntaxError: invalid syntax)
```

```python
#### Rule 5: It should not have hyphen in between

#first-name = 'Parag'(SyntaxError: cannot assign to expression here. Maybe you meant '==' instead of '='?)
```

```python
#### Rule 6: It should not used any special character like ($,%,&) etc.

#first_name$ = 'Parag'(SyntaxError: invalid syntax)
```

```python
##### Naming convention ---> It must have a snake case

first_name = 'Parag'
last_name  = 'Pujari'

print(first_name)
print(last_name)
```

```python
##### You can assign multiple variables with nultiple values in one statement

name, age, salary = 'Parag', 31, 53874

print(f'Name of the employee is {name}')
print(f'Age  of the employee is {age}')
print(f'Salary of the employee is {salary}')
```

```python
##### assign one value to multiple variables

a = b = c = 100

print(a)
print(b)
print(c)
```

```python
##### Reassignment

salary = 50000

print(f'Old salary is {salary}')

salary = 60000

print(f'New salary is {salary}')
```

```python
##### Swapping of the variables

a = 10
b = 20

print(f'Value of a before swapping is {a}')
print(f'Value of b before swapping is {b}')


a,b = b,a

print(f'Value of a after swapping is {a}')
print(f'Value of b after swapping is {b}')
```

```python
##### Constant -------> It is a value that cannot be changed during the entire execution of the program

PI = 3.142

print(f'Value of PI is {PI}')
```

30.  You have the following variables:

original_price = 1000

discount = 200

tax = 80

Calculate the final price using only variables and arithmetic.

The logic is:

Price after discount = original price - discount

Final price = price after discount + tax

```python
original_price = 1000

discount = 200

tax = 80


##### Calculate the price after discount
price_after_discount = original_price - discount
print(f'Price after discount is {price_after_discount}')

##### Calculate the final price
final_price = price_after_discount + tax
print(f'Final price is {final_price}')
```
