#### Data Type

1.  A data type determines the type of each element used in the Python program.

2.  There are nine data types used in Python program. They are as follows:---

    (a.)    Integer

    (b.)    Float

    (c.)    Boolean

    (d.)    String

    (e.)    Complex

    (f.)    List

    (g.)    Tuples

    (h.)    Sets

    (i.)    Dictionary


3.  Python is dynamically typed. The data type of each element in Python is not defined. The data type of each element is known automatically once the value is allocated to it.

4.  The same variable has different values at different runtime execution of the program. So the same variable can have different data types.

5.  It helps in determining the identity of each element in the program.


| Category | Data Types | Example             |
| -------- | ---------- | ------------------- |
| Numeric  | `int`      | `10`                |
| Numeric  | `float`    | `10.5`              |
| Numeric  | `complex`  | `2 + 3j`            |
| Boolean  | `bool`     | `True`              |
| Text     | `str`      | `"Python"`          |
| Sequence | `list`     | `[10, 20, 30]`      |
| Sequence | `tuple`    | `(10, 20, 30)`      |
| Set      | `set`      | `{10, 20, 30}`      |
| Mapping  | `dict`     | `{"name": "Parag"}` |
| Special  | `NoneType` | `None`              |




```python
##### 1. Integer data type

x = 10
print(x)
print(f'Data Type of x is {type(x)}')
```

```python
##### 2. Floating data type

x = 10.24

print(x)

print(f'Data type of x is {type(x)}')
```

```python
##### 3. Boolean data type

x = True

print(x)

print(type(x))
```

```python
##### 4. String data type

x = 'Parag'

print(x)

print(type(x))
```

```python
##### 5. Complex data type

x = 2+3j

print(x)

print(type(x))
```

```python
##### 6. List

x = [1,2,3,4,5]

print(x)

print(type(x))
```

```python
##### 7. Tuples

x = (1,2,3,4,5)

print(x)

print(type(x))
```

```python
##### 8. Sets

x = {1,2,2,3,3,3,4,5}

print(x)

print(type(x))
```

```python
##### 9. Dictionaries

x = {'name':'Parag', 'age':31,'salary':53874,'gender':'male'}

print(x)

print(type(x))
```
