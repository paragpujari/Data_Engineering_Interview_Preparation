##### Properties of Variable

(a.)  A Variable in Python is dynamically typed. The data type of the variable is determined automatically once the value is assigned to the variable.The value of the variable gets changed at different runtime execution of the program.

```python
x = 10
print(x)
print(type(x))


x = 80.98
print(x)
print(type(x))

x = 'Alice'
print(x)
print(type(x))

x = True
print(x)
print(type(x))

x = (3+4j)
print(x)
print(type(x))
```

##### OBSERVATIONS:

(a.)    The same variable x is initialized with different values of data at different runtime execution of the program.

##### Rules for defining the variable

1.  A Variable mush have letters, numbers or underscore.

```python
num = 10  ###### Variable with letters
num1 = 10 #####  Variable with number
num1_ans = 10 ##### Variable with letter, number or underscore

print(num)
print(num1)
print(num1_ans)
```

2.  A Variable must begin with letter or an underscore.

```python
num1  =  10       ##### A Variable beginning with letter
print(num1)

_num = 10         ##### A Variable beginning with underscore
print(_num)
```

3.  A Variable must have an underscore if needed

```python
first_name = 'Parag'

last_name  = 'Pujari'

print(first_name)

print(last_name)
```

4. A Variable should not begin with a digit.

```python
# 1num = 10
```

5. A Variable name should not be the name of the keyword.

```python
# def = 10
```

6. A Variable name should not have hyphen in between

```python
# first name = 'Parag'
```

7. A Variable name should not have any special symbols in them

```python
# $name = 'Parag'  
# %name = 'Parag'  
# @name = 'Parag' 
```
