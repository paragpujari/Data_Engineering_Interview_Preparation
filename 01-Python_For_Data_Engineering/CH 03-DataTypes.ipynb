##### Data Types

1.  Data Type is a type that helps in determining the type of each data element used in the Python program.

2.  It is not defined. It gets automatically determined once the value is assigned to it.

##### Main Inbuilt Data Types

| Category      |           Data Types              |
|---------------|-----------------------------------|
| Numeric       |   Integer, Float, Complex         |
| Boolean       |   bool                            |
| Text          |   str                             |
| Sequence      |   list, tuple, range              |
| Mapping       |   Dict                            |
| Set           |   set, frozenset                  |
| Binary        |   bytes, bytearray                |
| Special       |   NoneType                        |

##### Data Types

1.  Numeric ----->  It holds the numbers or the numeric values.

(a.)    Integer ----->  It holds the whole number.

```python
#####  Integer

x = 10

print(x)

print(type(x))
```

(b.) Float  ----->  It holds the decimal portion.

```python
##### Float

x = 90.76

print(x)

print(type(x))
```

(c.) Complex -----> It holds the real and imaginary part.

```python
x = (3+4j)

print(x)

print(type(x))
```

2.  Boolean ----->  It holds the value either True or False.

```python
x = True

print(x)

print(type(x))
```

```python
x = False

print(x)

print(type(x))
```

3. Text -----> It holds the textual information.

```python
name = 'Parag'

print(name)

print(type(name))
```

3. Sequence -----> It holds all the data in series.

(a.)    Lists -----> It contains all the elements enclosed within [].

```python
lists = [10,20,30,40,50]

print(lists)

print(type(lists))
```

(b.)    Tuples -----> It contains all the elements within ().

```python
tuples = (10,20,30,40,50)

print(tuples)

print(type(tuples))
```

(c.)    range   ----->  It contains the numbers in sequence

```python
ans = list(range(1,11))

print(ans)

print(type(ans))
```

4. Mapping  ----->  It helps in pairing key with value elements.

(a.)    Dictionary  ----->  It helos in pairing the key with value pairs.

```python
employee = {
    'emp_id'    :   472             ,
    'emp_name'  :   'Parag'         ,
    'emp_age'   :   31              ,
    'emp_gender':   'Male'          ,
    'emp_salary':   53874.78
}

print(employee)

print(type(employee))
```

5. Set -----> It has no duplicate elements.

(a.)    Set -----> It has no duplicate elements. It is enclosed within {}. It is defined using set()

```python
ans = set({1,2,2,2,3,3,3,4,4,5,5})

print(ans)

print(type(ans))
```

(b.) Frozenset -----> It is defined using frozenset(). Here we cannot perform any modifications in it

```python
ans = frozenset({1,2,3,4,4,5})

print(ans)

print(type(ans))
```

##### Difference between set and frosen set

|   set                                             |               frozenset                           |
|---------------------------------------------------|---------------------------------------------------|
| It is a mutuable collection of unique elements    | It is a immutuable collection of unique elements  |

6. Binary

(a.)    bytes ----->    It is an immutable sequence of elements representing the binary data.

```python
x = b'789'

print(x)

print(type(x))
```

```python
b = bytes([65, 66, 67])

print(b)
```

7. Special Data Type

(a.)    None -----> It is the missing or unknown value. It does not have any specific type.

```python
x = None

print(x)

print(type(x))
```
