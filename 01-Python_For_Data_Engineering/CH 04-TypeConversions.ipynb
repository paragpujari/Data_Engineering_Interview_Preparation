##### Type Conversions

1.  Type Conversion is a technique of converting the value of one data type into another.

##### Types of Type Conversions

(a.)    Implicit Type Conversions ----> It refers to the process where the data type of one value gets automatically converted into the another data type.

```python
x = 10
y = 5.7

print(f'Original value of x is {x}')

x = x + y
print(f'Converted value of x is {type(x)}')
```

(b.) Explicit Type Conversions -----> It refers to the technique of manually converting the data type of one value to another data type

```python
x=10
print(x)
print(f'Original data type of x is {type(x)}')

##### convert the data type of int to float
ans = float(x)
print(ans)
print(f'Converted data type of ans is {type(ans)}')
```

##### Properties of Type Conversion

(a.) On converting the value from lower data type (int) to higher data type(float), there is no loss of data. It is safe to use.

(b.) On converting the value from higher data type(float)  to lower data type(int), there is a loss of precision. It is not safe to use.

##### Common Type Conversion functions

|    Function	  | Conversion                                        |                           
|-----------------|---------------------------------------------------|
|int()	          | Converts into integer                             |
|float()	      | Converts into float                               |
|str()	          | Converts into string                              |

|list()	          | Converts into list                                |

|set()	          | Converts into set                                 |
|dict()	          | Converts suitable key-value pairs into dictionary |


##### Solved Examples

1.  Consider a age and convert from Integer to Float


```python
age = 31
print(age)
print(type(age))
##### convert the age from integer to float
ans = float(age)
print(ans)
print(type(ans))
```

2. Consider a salary and convert from float to integer

```python
salary = 53874.67
print(salary)
print(type(salary))
##### convert the salary from float to integer
ans = int(salary)
print(ans)
print(type(ans))
```

##### OBSERVATIONS:

(a.)    Here the salary is in floating data, on convert it to integer, theere is a loss of precision of(.67) that gets lost after performing the type conversion.

3. Consider a number and convert it from Integer to String

```python
x=10
print(x)
print(type(x))

##### convert the number from integer to string
ans = str(x)
print(ans)
print(type(ans))
```

4.  Consider a price and convert it from String to Integer

```python
price = '53476'
print(price)
print(type(price))

##### convert the price from string to integer
ans = int(price)
print(ans)
print(type(ans))
```

5. Consider a list and convert it from List to Tuple and Set

```python
product_list = ["Laptop","Smartphone","Tablet"]

##### convert from list -----> Tuple
product_tuple = tuple(product_list)
print(product_tuple)
print(type(product_tuple))

##### convert from tuple -----> Set
product_set = set(product_tuple)
print(product_set)
print(type(product_set))
```
